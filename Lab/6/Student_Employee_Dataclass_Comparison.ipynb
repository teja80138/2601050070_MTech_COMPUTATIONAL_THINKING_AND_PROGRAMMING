# Student/Employee Data Model

Comparison of a traditional class and dataclass implementation.

```python
from dataclasses import dataclass


# Traditional class
class Student:
    def __init__(self, id: int, name: str, marks: float):
        self.id = id
        self.name = name
        self.marks = marks


# Dataclasses
@dataclass
class StudentData:
    id: int
    name: str
    marks: float


@dataclass
class Employee:
    id: int
    name: str
    salary: float


# Create objects
s1 = Student(101, 'Teja', 85.5)
s2 = StudentData(102, 'Ravi', 90.0)
e1 = Employee(201, 'Kiran', 50000.0)

print('Traditional Class:')
print(s1.__dict__)

print('\nDataclass Student:')
print(s2)

print('\nDataclass Employee:')
print(e1)
```
