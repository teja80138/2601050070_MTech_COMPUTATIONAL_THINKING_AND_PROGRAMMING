# 0/1 Knapsack using Dynamic Programming

This notebook contains the Python implementation of the 0/1 Knapsack problem using Dynamic Programming.

````python
# 0/1 Knapsack using Dynamic Programming

def knapsack(weights, values, capacity):
    n = len(weights)

    # Create DP table
    dp = [[0 for _ in range(capacity + 1)]
          for _ in range(n + 1)]

    # Fill the DP table
    for i in range(1, n + 1):
        for w in range(1, capacity + 1):

            # If the current item cannot fit
            if weights[i - 1] > w:
                dp[i][w] = dp[i - 1][w]

            else:
                # Maximum of excluding or including the item
                dp[i][w] = max(
                    dp[i - 1][w],
                    values[i - 1] + dp[i - 1][w - weights[i - 1]]
                )

    return dp[n][capacity]


# Input data
weights = [2, 3, 4, 5]
values = [3, 4, 5, 6]
capacity = 5

# Calculate maximum value
maximum_value = knapsack(weights, values, capacity)

# Display result
print("Weights:", weights)
print("Values:", values)
print("Knapsack Capacity:", capacity)
print("Maximum Value:", maximum_value)
````
