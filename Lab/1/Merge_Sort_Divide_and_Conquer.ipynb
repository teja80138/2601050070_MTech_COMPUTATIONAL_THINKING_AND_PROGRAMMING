# Merge Sort using Divide-and-Conquer

## Objective
Implement Merge Sort using the Divide-and-Conquer technique and analyze its time complexity.

## Algorithm

1. Divide the array into two halves.
2. Recursively sort the left half.
3. Recursively sort the right half.
4. Merge the two sorted halves.
5. Repeat until the complete array is sorted.

```python
def merge(arr, left, mid, right):
    # Create temporary arrays for the two halves
    L = arr[left:mid + 1]
    R = arr[mid + 1:right + 1]

    i = 0
    j = 0
    k = left

    # Merge the two sorted halves
    while i < len(L) and j < len(R):
        if L[i] <= R[j]:
            arr[k] = L[i]
            i += 1
        else:
            arr[k] = R[j]
            j += 1
        k += 1

    # Copy remaining elements from the left half
    while i < len(L):
        arr[k] = L[i]
        i += 1
        k += 1

    # Copy remaining elements from the right half
    while j < len(R):
        arr[k] = R[j]
        j += 1
        k += 1


def merge_sort(arr, left, right):
    # Divide and conquer
    if left < right:
        mid = (left + right) // 2

        # Sort left half
        merge_sort(arr, left, mid)

        # Sort right half
        merge_sort(arr, mid + 1, right)

        # Combine the sorted halves
        merge(arr, left, mid, right)
```

```python
# Main program
arr = [38, 27, 43, 3, 9, 82, 10]

print("Original array:")
print(arr)

merge_sort(arr, 0, len(arr) - 1)

print("Sorted array:")
print(arr)
```

## Output

Original array:
`[38, 27, 43, 3, 9, 82, 10]`

Sorted array:
`[3, 9, 10, 27, 38, 43, 82]`

## Time Complexity

The recurrence relation is:

`T(n) = 2T(n/2) + O(n)`

- Best Case: **O(n log n)**
- Average Case: **O(n log n)**
- Worst Case: **O(n log n)**
- Space Complexity: **O(n)**

Merge Sort divides the array into halves for `log n` levels, and each level requires `O(n)` time for merging. Therefore, the overall time complexity is **O(n log n)**.
