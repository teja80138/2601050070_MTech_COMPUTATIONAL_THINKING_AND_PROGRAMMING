# Banking Management System
Demonstrates abstraction, inheritance, polymorphism, dataclasses, and type hints.

```python
from abc import ABC, abstractmethod
from dataclasses import dataclass

@dataclass
class BankAccount(ABC):
    no: int
    name: str
    balance: float

    def deposit(self, amount: float) -> None:
        self.balance += amount

    @abstractmethod
    def withdraw(self, amount: float) -> None:
        pass

@dataclass
class Savings(BankAccount):
    def withdraw(self, amount: float) -> None:
        if self.balance - amount >= 500:
            self.balance -= amount
        else:
            print('Minimum balance required')

@dataclass
class Current(BankAccount):
    def withdraw(self, amount: float) -> None:
        if self.balance - amount >= -1000:
            self.balance -= amount
        else:
            print('Overdraft limit exceeded')

s = Savings(101, 'Teja', 5000)
c = Current(102, 'Ravi', 2000)

s.deposit(1000)
s.withdraw(2000)
c.deposit(500)
c.withdraw(2500)

print(s)
print(c)
```
