# Stack and Queue using Type Hints and Dataclasses


```python
from dataclasses import dataclass, field
from collections import deque
from typing import TypeVar, Generic

T = TypeVar('T')

@dataclass
class Stack(Generic[T]):
    items: list[T] = field(default_factory=list)

    def push(self, x: T):
        self.items.append(x)

    def pop(self) -> T:
        return self.items.pop()

    def peek(self) -> T:
        return self.items[-1]


@dataclass
class Queue(Generic[T]):
    items: deque[T] = field(default_factory=deque)

    def enqueue(self, x: T):
        self.items.append(x)

    def dequeue(self) -> T:
        return self.items.popleft()

    def peek(self) -> T:
        return self.items[0]


# Stack
s = Stack[int]()
s.push(10)
s.push(20)
s.push(30)

print('Stack:', s.items)
print('Pop:', s.pop())
print('Peek:', s.peek())

# Queue
q = Queue[str]()
q.enqueue('A')
q.enqueue('B')
q.enqueue('C')

print('\nQueue:', list(q.items))
print('Dequeue:', q.dequeue())
print('Peek:', q.peek())
```
