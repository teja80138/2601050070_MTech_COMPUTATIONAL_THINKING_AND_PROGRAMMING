# List-Based vs Generator-Based Processing

Compare list-based and generator-based processing for a large dataset using execution time and memory usage.

```python
import time
import tracemalloc

N = 1_000_000


# List-based processing
def list_processing(n):
    data = [i for i in range(n)]
    return [i * i for i in data]


# Generator-based processing
def generator_processing(n):
    for i in range(n):
        yield i * i


# List test
tracemalloc.start()
start = time.time()
result = list_processing(N)
list_time = time.time() - start
_, list_memory = tracemalloc.get_traced_memory()
tracemalloc.stop()


# Generator test
tracemalloc.start()
start = time.time()
total = sum(generator_processing(N))
generator_time = time.time() - start
_, generator_memory = tracemalloc.get_traced_memory()
tracemalloc.stop()


# Display results
print("List-Based Processing")
print("Execution Time:", round(list_time, 4), "seconds")
print("Peak Memory:", round(list_memory / (1024 * 1024), 2), "MB")

print("\nGenerator-Based Processing")
print("Execution Time:", round(generator_time, 4), "seconds")
print("Peak Memory:", round(generator_memory / (1024 * 1024), 2), "MB")
```
