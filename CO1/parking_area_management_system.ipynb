# Parking Area Management System

A Python-based parking management system for 100 parking slots. It supports availability checking, slot allocation, vehicle exit, parking charges, vehicle search, and full-parking detection.

```python
# ==========================================
# PARKING AREA MANAGEMENT SYSTEM
# ==========================================

TOTAL_SLOTS = 100

# Parking slots
# Key   -> Slot number
# Value -> Vehicle number or None
parking_slots = {
    slot: None
    for slot in range(1, TOTAL_SLOTS + 1)
}

# Store vehicle information
vehicles = {}


# ==========================================
# DISPLAY PARKING AVAILABILITY
# ==========================================

def show_availability():

    available_slots = [
        slot
        for slot, vehicle in parking_slots.items()
        if vehicle is None
    ]

    occupied_slots = [
        slot
        for slot, vehicle in parking_slots.items()
        if vehicle is not None
    ]

    print("\n========== PARKING AVAILABILITY ==========")

    print(f"Total Slots    : {TOTAL_SLOTS}")
    print(f"Occupied Slots : {len(occupied_slots)}")
    print(f"Available Slots: {len(available_slots)}")

    if available_slots:
        print("\nAvailable Slot Numbers:")
        print(available_slots)
    else:
        print("\nPARKING AREA IS FULL!")

    print("===========================================")


# ==========================================
# PARKING AREA FULL CHECK
# ==========================================

def is_parking_full():

    return all(
        vehicle is not None
        for vehicle in parking_slots.values()
    )


# ==========================================
# ALLOCATE PARKING SLOT
# ==========================================

def allocate_slot():

    if is_parking_full():

        print("\nParking area is FULL.")
        print("No parking slots are available.")
        return

    vehicle_number = input(
        "\nEnter vehicle number: "
    ).strip().upper()

    if not vehicle_number:
        print("Vehicle number cannot be empty.")
        return

    # Check if vehicle is already parked
    if vehicle_number in vehicles:

        print(
            f"Vehicle {vehicle_number} is already parked."
        )
        return

    # Find first available slot
    for slot in parking_slots:

        if parking_slots[slot] is None:

            parking_slots[slot] = vehicle_number

            vehicles[vehicle_number] = {
                "slot": slot
            }

            print("\n========== SLOT ALLOCATED ==========")
            print(f"Vehicle Number : {vehicle_number}")
            print(f"Parking Slot   : {slot}")
            print("====================================")

            return


# ==========================================
# RELEASE PARKING SLOT
# ==========================================

def release_slot():

    vehicle_number = input(
        "\nEnter vehicle number leaving: "
    ).strip().upper()

    if vehicle_number not in vehicles:

        print(
            "Vehicle not found in the parking area."
        )
        return

    slot = vehicles[vehicle_number]["slot"]

    # Get parking duration
    while True:

        try:

            hours = float(
                input(
                    "Enter number of hours parked: "
                )
            )

            if hours <= 0:

                print(
                    "Hours must be greater than 0."
                )

            else:
                break

        except ValueError:

            print(
                "Please enter a valid number."
            )

    # Calculate parking charges
    charges = calculate_charges(hours)

    # Release slot
    parking_slots[slot] = None

    del vehicles[vehicle_number]

    print("\n========== VEHICLE EXIT ==========")
    print(f"Vehicle Number : {vehicle_number}")
    print(f"Parking Slot   : {slot}")
    print(f"Hours Parked   : {hours:.2f}")
    print(f"Parking Charge : ₹{charges:.2f}")
    print("Slot Released Successfully.")
    print("==================================")


# ==========================================
# CALCULATE PARKING CHARGES
# ==========================================

def calculate_charges(hours):

    # First hour = ₹30
    # Every additional hour = ₹20

    if hours <= 1:

        charge = 30

    else:

        additional_hours = hours - 1

        charge = 30 + (additional_hours * 20)

    return charge


# ==========================================
# DISPLAY ALL PARKED VEHICLES
# ==========================================

def show_parked_vehicles():

    print("\n========== PARKED VEHICLES ==========")

    parked = False

    for slot, vehicle in parking_slots.items():

        if vehicle is not None:

            parked = True

            print(
                f"Slot {slot:<5} -> {vehicle}"
            )

    if not parked:

        print("No vehicles are currently parked.")

    print("=====================================")


# ==========================================
# FIND VEHICLE
# ==========================================

def find_vehicle():

    vehicle_number = input(
        "\nEnter vehicle number to search: "
    ).strip().upper()

    if vehicle_number in vehicles:

        slot = vehicles[vehicle_number]["slot"]

        print("\n========== VEHICLE FOUND ==========")
        print(f"Vehicle Number : {vehicle_number}")
        print(f"Parking Slot   : {slot}")
        print("===================================")

    else:

        print(
            "Vehicle is not currently parked."
        )


# ==========================================
# MAIN MENU
# ==========================================

while True:

    print("\n\n========================================")
    print("       PARKING AREA MANAGEMENT SYSTEM")
    print("========================================")

    print("1. Show Parking Availability")
    print("2. Allocate Slot to Vehicle")
    print("3. Release Slot / Vehicle Exit")
    print("4. Show Parked Vehicles")
    print("5. Find Vehicle")
    print("6. Check if Parking is Full")
    print("7. Exit")

    print("========================================")

    choice = input(
        "Enter your choice: "
    ).strip()

    # 1. Show availability
    if choice == "1":

        show_availability()

    # 2. Allocate slot
    elif choice == "2":

        allocate_slot()

    # 3. Release slot
    elif choice == "3":

        release_slot()

    # 4. Show parked vehicles
    elif choice == "4":

        show_parked_vehicles()

    # 5. Find vehicle
    elif choice == "5":

        find_vehicle()

    # 6. Check full
    elif choice == "6":

        if is_parking_full():

            print(
                "\nPARKING AREA IS FULL!"
            )

        else:

            available = sum(
                1
                for vehicle in parking_slots.values()
                if vehicle is None
            )

            print(
                "\nParking is not full."
            )

            print(
                f"Available slots: {available}"
            )

    # 7. Exit
    elif choice == "7":

        print(
            "\nThank you for using the "
            "Parking Area Management System!"
        )

        break

    else:

        print(
            "\nInvalid choice. Please try again."
        )
```

## Parking Charges

- First hour: ₹30
- Every additional hour: ₹20

Example for 3 hours: **₹30 + (2 × ₹20) = ₹70**.

## Main Features

1. Show available slots
2. Allocate a slot to a vehicle
3. Release a slot when a vehicle leaves
4. Calculate parking charges
5. Show parked vehicles
6. Find a vehicle
7. Identify when the parking area is full
